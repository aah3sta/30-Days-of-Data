```python
#PYTHON LISTS - use square brackets
programs = ["KamiLimu", "ABF", "Mozilla Buildathon", "30-Days-of-Data"]
print (programs)

print(programs[2]) #accessing the 3rd element of the list

#negative indexes
print(programs[-1]) #accessing the last element of the list

#range of values
print(programs[0:2]) #all values between index 0 and 2 (not including 2)

#slicing
print(programs[:3]) #all values from 0 to 3, not including 3
print(programs[1:]) #all values from index 1 to the end of the list

#adding item list
# using append - adds to end of a list
programs.append("Hevy Hub")
print(programs)

#using insert - adds to a specific index
programs.insert(1, "Python for Data Science")
print(programs)

#using extend - adds multiple items to a list
programs_2 = ["Data Science with Python", "Data Science with R"]
programs.extend(programs_2)
print(programs)

#removing items from a list
#using remove - removes a specific item from the list
programs.remove("Data Science with R")
print(programs)

#using pop - removes an item at a specific index
programs.pop(-1)
print(programs)

#Sorting a list
#reverse
programs.reverse()
print(programs)

#sort in alphabetical or ascending order
programs.sort()
print(programs)

#sorting in descendig order
programs.sort(reverse=True)
print(programs)

sorted(programs)
print(programs) #original list remains unchanged
```

```python
#minimum and maximum values in a list
numbers = [1, 2, 3, 4, 5]

numbers.sort(reverse=True)
print("Reversed:", numbers)

#minimum value
print("Minimum value:", min(numbers))

#maximum value
print("Maximum value:", max(numbers))

#sum
print("Sum:", sum(numbers))

#finding values in list
print(programs.index("ABF"))

#Is the value in the list?
print("KamiLimu" in programs) #true to show that it is

#printing all items in the list
for number in numbers:
    print(number)

#enumerate prints index plus value    
for index, program in enumerate(programs): 
    print(index, program)

#turning list into string
program_str = ", ".join(programs)
print (program_str)
```

```python
#TUPLES - use parentheses ()
#are immutable

subjects = ("Math", "Physics", "English")
print (subjects)
```

```python
#SETS - unordered, have no duplicates. Use curly brackets

sciences = {"Biology", "Chemistry", "Physics", "Biology"}
print(sciences)

print ("Art" in sciences)

#intersection - what is repeated in both
print(sciences.intersection(subjects))

#difference - what isnt shared
print(sciences.difference(subjects))

#union - all together
print(sciences.union(subjects))
```

```python
#DICTIONARIES - key value pairs / hash maps
student = {"name" : "John",
           "age" : 25,
           "courses" : ["Math", "CS"]}

print(student["name"])

#use get method to return none in case data doesnt exist
print(student.get("Year"))

#updating Value
student["name"] = "Esther"

#update method
student.update({"age" : 20, "phone" : "0712345678"})

#deleting
del student["phone"]

#adding new entry to dictionary
student["year"] = 4
print(student)

#length - no of keys
print(len(student))

#looping through dictionary
for key, value in student.items():
    print(key, value)

```

```python
#ASSIGNMENT
""" Create a dictionary representing 5 e-commerce transactions with keys user_id, items (list of strings), and amount. 
 Write a Python script to calculate the total spent per unique user_id and find the most frequently purchased item. """
```

```python
transactions = [{"user_id": "001", "items": ["pencil", "eraser", "sharpener"], "amount": [50]},
                {"user_id": "002", "items": ["pencil", "book", "sharpener"], "amount": [90]},
                {"user_id": "001", "items": ["sharpener"], "amount": [20]},
                {"user_id": "004", "items": ["pencil", "book", "sharpener", "eraser"], "amount": [100]},
                {"user_id": "002", "items": ["pencil", "book", "sharpener", "eraser"], "amount": [100]}]

print("--- TRANSACTIONS ---")
for transaction in transactions:
    print(transaction)

#total spent per unique user id
#create a dictionary to hold the user_id and total amount mappings
total_spent = {}

for transaction in transactions:
    user = transaction["user_id"]
    amount = transaction["amount"]

    if user in total_spent:
        total_spent[user] += amount
    else:
        total_spent[user] = amount

print("\n--- TOTAL SPENT PER USER ---")
for user_id, amount in total_spent.items():
    print(f"User ID {user_id}: ${amount}")

#most frequently purchased
item_counts = {}

for transaction in transactions:
    for item in transaction["items"]:
        # dict.get(item, 0) returns the current count or 0 if item doesn't exist yet
        item_counts[item] = item_counts.get(item, 0) + 1

print("\n--- ITEM PURCHASE FREQUENCIES ---")
for item, count in item_counts.items():
    print(f"{item.capitalize()}: {count} time(s)")

# Find the item with the highest frequency count
most_popular_item = max(item_counts, key=item_counts.get)
max_count = item_counts[most_popular_item]

print("\n--- FINAL SUMMARY ---")
print(f"Most Popular Item: '{most_popular_item}' (Bought {max_count} times)")
```
